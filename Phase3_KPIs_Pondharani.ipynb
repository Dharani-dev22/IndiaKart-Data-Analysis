```python
import pandas as pd

orders = pd.read_csv('orders DA.csv')
customers = pd.read_csv('customers DA.csv')

orders['order_date'] = pd.to_datetime(orders['order_date'], format='%d-%m-%Y')
```

```python
gmv = orders['final_amount'].sum()

delivered_orders = orders[orders['status'] == 'Delivered']
net_revenue = delivered_orders['final_amount'].sum()

aov = net_revenue / len(delivered_orders)

print(f"Gross Merchandise Value (GMV): {gmv:,.2f}")
print(f"Net Revenue: {net_revenue:,.2f}")
print(f"Average Order Value (AOV): {aov:,.2f}")
```

```python
returns_df = pd.read_csv('returns DA.csv')

cancelled_orders = orders[orders['status'] == 'Cancelled']
cancellation_rate = (len(cancelled_orders) / len(orders)) * 100

return_rate = (len(returns_df) / len(delivered_orders)) * 100

print(f"Cancellation Rate: {cancellation_rate:.2f}%")
print(f"Return Rate: {return_rate:.2f}%")
```

```python
payments = pd.read_csv('payments DA.csv')

clv = customers.groupby('segment')['total_spent'].mean()

failed_payments = len(payments[payments['status'] == 'Failed'])
payment_failure_rate = (failed_payments / len(payments)) * 100

print("Customer Lifetime Value (CLV) by Segment:")
print(clv.apply(lambda x: f"Rs.{x:,.0f}"))
print(f"\nPayment Failure Rate: {payment_failure_rate:.2f}%")
```

```python
order_items = pd.read_csv('order_items DA.csv')
inventory = pd.read_csv('inventory DA.csv')

monthly_gmv = orders.groupby(orders['order_date'].dt.to_period('M'))['final_amount'].sum()
mom_growth = monthly_gmv.pct_change() * 100

cat_revenue = order_items.groupby('category')['total_price'].sum()
top_category = cat_revenue.idxmax()
top_cat_share = (cat_revenue.max() / cat_revenue.sum()) * 100

in_stock = len(inventory[inventory['quantity_available'] > 0])
inventory_fill_rate = (in_stock / len(inventory)) * 100

print(f"Top Category: {top_category} ({top_cat_share:.2f}% Revenue Share)")
print(f"Inventory Fill Rate: {inventory_fill_rate:.2f}%\n")
print("Month-over-Month Growth (Last 3 Months):")
print(mom_growth.tail(3).apply(lambda x: f"{x:.2f}%"))
```
